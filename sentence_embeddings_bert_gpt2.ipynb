# Experiment 25: Sentence Embeddings using BERT and GPT-2

**Aim:** Create sentence representations with a BERT encoder and a GPT-2 hidden-state model.

**Libraries:** `transformers, torch`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
# pip install transformers torch
import torch
from transformers import AutoTokenizer, AutoModel
sentences=["Natural language processing is fascinating.","Transformers create contextual representations."]
def mean_pool(hidden, mask):
    mask=mask.unsqueeze(-1).expand(hidden.size()).float()
    return (hidden*mask).sum(1)/mask.sum(1).clamp(min=1e-9)
for name in ["bert-base-uncased","gpt2"]:
    tokenizer=AutoTokenizer.from_pretrained(name)
    if tokenizer.pad_token is None: tokenizer.pad_token=tokenizer.eos_token
    model=AutoModel.from_pretrained(name)
    encoded=tokenizer(sentences,padding=True,truncation=True,return_tensors="pt")
    with torch.no_grad(): output=model(**encoded)
    embeddings=mean_pool(output.last_hidden_state,encoded["attention_mask"])
    print(name,"embedding shape:",tuple(embeddings.shape))
    print("First vector (first 8 values):",embeddings[0,:8].numpy().round(3))
```
