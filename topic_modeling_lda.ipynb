# Experiment 17: Topic Modeling using LDA

**Aim:** Discover latent topics in a small collection of documents using LDA.

**Libraries:** `scikit-learn`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
docs = ["football team scored goal match player", "team won football league match", "neural networks train machine learning model", "machine learning uses neural network data", "python code trains a learning model", "football player kicked the ball"]
vectorizer = CountVectorizer(stop_words='english')
X = vectorizer.fit_transform(docs)
lda = LatentDirichletAllocation(n_components=2, random_state=42, max_iter=20)
lda.fit(X)
terms = vectorizer.get_feature_names_out()
for i, topic in enumerate(lda.components_):
    top = topic.argsort()[-5:][::-1]
    print(f"Topic {i+1}:", [terms[j] for j in top])
print("Document-topic distribution:\n", lda.transform(X).round(2))
```
