# Experiment 27: Named Entity Recognition using Transformers

**Aim:** Run token-classification NER with a pretrained Hugging Face model.

**Libraries:** `transformers, torch`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
# pip install transformers torch
from transformers import pipeline
ner=pipeline("token-classification",model="dslim/bert-base-NER",aggregation_strategy="simple")
text="Ankit Jha studies at NIET in Greater Noida and visited Google in California."
for entity in ner(text):
    print(entity["word"],entity["entity_group"],round(entity["score"],3))
```
