# Experiment 23: Implementing a Basic Attention Mechanism

**Aim:** Calculate attention weights and a context vector from query and key-value vectors.

**Libraries:** `numpy`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
import numpy as np
np.random.seed(42)
# Scaled dot-product attention: softmax(Q K^T / sqrt(d_k)) V
Q=np.random.rand(1,4); K=np.random.rand(3,4); V=np.random.rand(3,5)
scores=(Q @ K.T)/np.sqrt(K.shape[-1])
weights=np.exp(scores-scores.max(axis=-1,keepdims=True))
weights=weights/weights.sum(axis=-1,keepdims=True)
context=weights @ V
print("Attention scores:", scores)
print("Attention weights:", weights)
print("Context vector:", context)
print("Weights sum to:", weights.sum())
```
