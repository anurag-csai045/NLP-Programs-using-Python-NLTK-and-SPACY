```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
#sample corpus (collection of sentences)
corpus=[ "I love ML",
"I love coding in python",
"ML is amazing",
"python is great for ML"]
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer()
X_tfidf= tfidf.fit_transform(corpus) # Use tfidf to fit and transform
tfidf_df=pd.DataFrame(X_tfidf.toarray(),columns=tfidf.get_feature_names_out())
tfidf_df
```

```python
vectorizer = CountVectorizer()
X_bow = vectorizer.fit_transform(corpus)
bow_df = pd.DataFrame(X_bow.toarray(), columns=vectorizer.get_feature_names_out())
print("Bag-of-Words DataFrame:")
display(bow_df)
```

Now, let's combine the TF-IDF and Bag-of-Words DataFrames.

```python
combined_df = pd.concat([tfidf_df, bow_df], axis=1)
print("Combined TF-IDF and Bag-of-Words DataFrame:")
display(combined_df)
```
