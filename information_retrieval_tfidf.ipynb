# Experiment 21: Information Retrieval with TF-IDF Ranking

**Aim:** Rank a small document collection according to a user query.

**Libraries:** `scikit-learn`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
documents = ["Python is used for data science and machine learning", "Football is played by two teams", "Natural language processing analyzes text", "Machine learning is a branch of artificial intelligence"]
query = "machine learning and artificial intelligence"
vectorizer=TfidfVectorizer(stop_words='english')
D=vectorizer.fit_transform(documents)
q=vectorizer.transform([query])
scores=cosine_similarity(q,D).ravel()
for idx in scores.argsort()[::-1]:
    print(f"Score={scores[idx]:.3f} | Document {idx+1}: {documents[idx]}")
```
