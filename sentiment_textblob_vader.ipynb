# Experiment 16: Sentiment Analysis using TextBlob and VADER

**Aim:** Estimate polarity and sentiment labels with two lexicon-based tools.

**Libraries:** `textblob, vaderSentiment`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
# Install if needed: pip install textblob vaderSentiment
from textblob import TextBlob
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
reviews = ["I absolutely love this product!", "The service was okay, nothing special.", "This is a terrible experience."]
analyzer = SentimentIntensityAnalyzer()
for review in reviews:
    polarity = TextBlob(review).sentiment.polarity
    vader = analyzer.polarity_scores(review)
    label = "Positive" if vader['compound'] >= .05 else "Negative" if vader['compound'] <= -.05 else "Neutral"
    print("\nReview:", review)
    print("TextBlob polarity:", round(polarity,3), "| VADER:", vader, "| Label:", label)
```
