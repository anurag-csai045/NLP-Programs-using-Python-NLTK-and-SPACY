# Experiment 14: Text Similarity using Word Mover’s Distance

**Aim:** Compute Word Mover’s Distance with pretrained word vectors.

**Libraries:** `gensim, scipy`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
# WMD requires a pretrained Word2Vec/KeyedVectors model. Download a compatible model first.
from pathlib import Path
from gensim.models import KeyedVectors
model_path = Path("../data/GoogleNews-vectors-negative300.bin")
if model_path.exists():
    keyed_vectors = KeyedVectors.load_word2vec_format(str(model_path), binary=True)
    doc1 = "machine learning understands text".lower().split()
    doc2 = "computers process language".lower().split()
    print("Word Mover's Distance:", keyed_vectors.wmdistance(doc1, doc2))
else:
    print("Model not included because it is very large. Add a compatible Word2Vec binary model to data/.")
```
