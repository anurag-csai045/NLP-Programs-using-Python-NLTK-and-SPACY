# Experiment 11: Cosine Similarity between Documents

**Aim:** Measure document similarity using TF-IDF vectors and cosine similarity.

**Libraries:** `scikit-learn`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
documents = ["I enjoy machine learning", "Machine learning is interesting", "I like cooking"]
X = TfidfVectorizer().fit_transform(documents)
similarity = cosine_similarity(X)
print(similarity.round(3))
print("Similarity of document 1 and 2:", round(similarity[0,1],3))
```
