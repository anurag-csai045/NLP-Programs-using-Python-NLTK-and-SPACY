# Experiment 31: Automatic Document Classification and Separation using ML

**Aim:** Classify sample documents into categories and separate them by predicted class.

**Libraries:** `scikit-learn`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
samples=["team scored goal in football match","player won the league","python code and software development","database and programming language","neural network trains on image data","machine learning model predicts values"]
labels=["Sports","Sports","Technology","Technology","AI","AI"]
model=Pipeline([("tfidf",TfidfVectorizer()),("clf",LinearSVC())]).fit(samples,labels)
new_docs=["football player scored a goal","python programming project","deep learning neural network"]
pred=model.predict(new_docs)
separated={label:[] for label in sorted(set(labels))}
for doc,label in zip(new_docs,pred): separated[label].append(doc)
for label,docs in separated.items(): print("\n",label,":",docs)
```
