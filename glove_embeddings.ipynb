# Experiment 13: GloVe Embeddings Loading and Vector Representation

**Aim:** Load pretrained GloVe vectors when available and inspect a word representation.

**Libraries:** `gensim (optional), numpy`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
# Download a GloVe text file separately (e.g. glove.6B.50d.txt) and place it in data/.
from pathlib import Path
import numpy as np
path = Path("../data/glove.6B.50d.txt")

def load_glove(path, limit=50000):
    vectors = {}
    with open(path, encoding="utf8") as f:
        for i, line in enumerate(f):
            if i >= limit: break
            values = line.rstrip().split()
            vectors[values[0]] = np.asarray(values[1:], dtype="float32")
    return vectors
if path.exists():
    embeddings = load_glove(path)
    word = "language"
    print("Vector dimension:", len(embeddings[word]))
    print(embeddings[word][:10])
else:
    print("Place glove.6B.50d.txt in the repository's data/ folder to run this practical.")
```
