# Experiment 20: Information Extraction from Structured and Unstructured Documents

**Aim:** Extract simple fields from structured records and entities from free text.

**Libraries:** `spacy`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
# Structured extraction from dictionaries / JSON-like records
records = [{"name":"Asha","role":"Data Analyst","city":"Delhi"},{"name":"Kabir","role":"ML Engineer","city":"Pune"}]
for record in records:
    print(f"{record['name']} works as {record['role']} in {record['city']}.")
```

```python
# Unstructured extraction with spaCy NER (trained model required)
import spacy
text = "OpenAI announced a new research project in San Francisco in 2025."
try:
    doc=spacy.load("en_core_web_sm")(text)
    print([(ent.text,ent.label_) for ent in doc.ents])
except OSError:
    print("Install model: python -m spacy download en_core_web_sm")
```
