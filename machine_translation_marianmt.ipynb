# Experiment 28: Machine Translation using MarianMT

**Aim:** Translate text using a pretrained English-to-French MarianMT model.

**Libraries:** `transformers, sentencepiece, torch`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
# pip install transformers sentencepiece torch
from transformers import MarianMTModel, MarianTokenizer
model_name="Helsinki-NLP/opus-mt-en-fr"
tokenizer=MarianTokenizer.from_pretrained(model_name)
model=MarianMTModel.from_pretrained(model_name)
texts=["Hello, how are you?", "Natural language processing is useful."]
batch=tokenizer(texts,return_tensors="pt",padding=True,truncation=True)
translated=model.generate(**batch,max_new_tokens=60)
for source, result in zip(texts,translated):
    print("English:",source,"\nFrench:",tokenizer.decode(result,skip_special_tokens=True))
```
