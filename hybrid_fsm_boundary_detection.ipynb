# Experiment 32: Hybrid Probabilistic + FSM-Based Sequence Modeling for Document Boundary Detection

**Aim:** Combine a simple probabilistic boundary score with a finite-state machine to segment a sequence into documents.

**Libraries:** `numpy`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
# Toy sequence: each item is a paragraph represented by lexical/formatting clues.
paragraphs=[
    {"text":"Quarterly sales increased by 12 percent.","title":False,"date":False,"topic_shift":False},
    {"text":"Revenue was strongest in the north region.","title":False,"date":False,"topic_shift":False},
    {"text":"Research Report: Language Models","title":True,"date":True,"topic_shift":True},
    {"text":"The experiment compares two transformer models.","title":False,"date":False,"topic_shift":False},
    {"text":"Appendix A: Evaluation Results","title":True,"date":False,"topic_shift":True}]
# Probability is a transparent weighted heuristic standing in for a trained boundary classifier.
def boundary_probability(p):
    score=0.10 + 0.45*p["title"] + 0.25*p["date"] + 0.30*p["topic_shift"]
    return min(score,0.99)
# FSM states: IN_DOCUMENT and NEW_DOCUMENT
state="IN_DOCUMENT"; documents=[[]]
for i,p in enumerate(paragraphs):
    prob=boundary_probability(p)
    if i>0 and state=="IN_DOCUMENT" and prob>=0.60:
        state="NEW_DOCUMENT"
    if state=="NEW_DOCUMENT":
        documents.append([]); state="IN_DOCUMENT"
    documents[-1].append(p["text"])
    print(f"Paragraph {i+1}: P(boundary)={prob:.2f}; state={state}")
print("\nDetected documents:")
for i,doc in enumerate(documents,1): print(f"Document {i}:",doc)
```
