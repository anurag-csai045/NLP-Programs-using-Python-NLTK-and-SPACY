# Experiment 15: Text Classification using Naive Bayes and SVM with TF-IDF

**Aim:** Train and compare Multinomial Naive Bayes and linear SVM classifiers.

**Libraries:** `scikit-learn`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.metrics import classification_report, accuracy_score
texts = ["excellent product and great quality", "I love this service", "very useful and reliable", "amazing experience", "terrible product and poor quality", "I hate this service", "not useful and disappointing", "awful experience", "great support", "bad support", "highly recommend", "never buy this"]
labels = ["positive"]*6 + ["negative"]*6
X_train, X_test, y_train, y_test = train_test_split(texts, labels, test_size=.33, random_state=42, stratify=labels)
for name, clf in [("Naive Bayes",MultinomialNB()),("Linear SVM",LinearSVC())]:
    pipe = Pipeline([("tfidf",TfidfVectorizer()),("classifier",clf)])
    pipe.fit(X_train,y_train); pred=pipe.predict(X_test)
    print(name, "accuracy:", accuracy_score(y_test,pred))
    print(classification_report(y_test,pred,zero_division=0))
```
