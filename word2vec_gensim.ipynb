# Experiment 12: Word2Vec Embeddings using Gensim

**Aim:** Train a small Word2Vec model on a custom corpus and inspect word vectors.

**Libraries:** `gensim`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
# If required: pip install gensim
from gensim.models import Word2Vec
sentences = [
    ["nlp","uses","machine","learning"], ["machine","learning","models","text"],
    ["word","embeddings","represent","words"], ["nlp","models","understand","language"],
    ["python","supports","nlp","projects"]
]
model = Word2Vec(sentences, vector_size=50, window=3, min_count=1, workers=1, seed=42, epochs=100)
print("Vector for 'nlp':", model.wv['nlp'][:10])
print("Most similar to 'machine':", model.wv.most_similar('machine', topn=3))
```
