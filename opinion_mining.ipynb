# Experiment 19: Opinion Mining on Product/Service Reviews

**Aim:** Extract review polarity and summarize positive/negative opinions.

**Libraries:** `textblob, pandas`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
import pandas as pd
from textblob import TextBlob
reviews = [
    ("Phone battery lasts all day and camera is excellent", "Phone"),
    ("The display is beautiful but the battery is poor", "Phone"),
    ("Delivery was fast and customer support was helpful", "Service"),
    ("The service was slow and staff were rude", "Service"),
    ("Good value for money", "Phone")]
rows=[]
for review, product in reviews:
    score=TextBlob(review).sentiment.polarity
    rows.append({"product":product,"review":review,"polarity":round(score,3),"opinion":"Positive" if score>.05 else "Negative" if score<-.05 else "Neutral"})
df=pd.DataFrame(rows); display(df)
print("Opinion summary:"); display(df.groupby(["product","opinion"]).size().unstack(fill_value=0))
```
