# N-Gram Extension: Bigrams and Trigrams
This supplementary notebook completes Practical 10 without modifying the original lab notebook. It demonstrates unigram, bigram, and trigram generation using scikit-learn.

```python
from sklearn.feature_extraction.text import CountVectorizer

corpus = [
    "Natural language processing is interesting",
    "Natural language processing uses machine learning",
    "Machine learning helps computers understand language"
]
```

```python
for label, ngram_range in [("Unigrams", (1, 1)), ("Bigrams", (2, 2)), ("Trigrams", (3, 3))]:
    vectorizer = CountVectorizer(ngram_range=ngram_range)
    matrix = vectorizer.fit_transform(corpus)
    print(f"\n{label}:")
    print(vectorizer.get_feature_names_out().tolist())
    print("Count matrix:")
    print(matrix.toarray())
```
