# Experiment 18: Topic Modeling using Latent Semantic Analysis

**Aim:** Use truncated SVD on a TF-IDF matrix to identify latent semantic topics.

**Libraries:** `scikit-learn`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
docs = ["football players score goals in a match", "the team won the football game", "machine learning models use training data", "deep learning is a machine learning method", "players train before the football match"]
vectorizer = TfidfVectorizer(stop_words='english')
X = vectorizer.fit_transform(docs)
lsa = TruncatedSVD(n_components=2, random_state=42).fit(X)
terms = vectorizer.get_feature_names_out()
for i, component in enumerate(lsa.components_):
    print(f"Topic {i+1}:", [terms[j] for j in component.argsort()[-5:][::-1]])
print("Document coordinates:\n", lsa.transform(X).round(3))
```
