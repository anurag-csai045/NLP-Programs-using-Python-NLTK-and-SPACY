# Experiment 22: Sequence Classification using RNN/LSTM

**Aim:** Build a compact LSTM sentiment classifier using a small illustrative dataset.

**Libraries:** `tensorflow`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
# This practical needs TensorFlow. Install with: pip install tensorflow
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
texts=["I love this movie", "This film is wonderful", "Amazing acting and story", "I hate this movie", "This film is boring", "Terrible acting and story"]
labels=[1,1,1,0,0,0]
tokenizer=Tokenizer(num_words=100, oov_token="<OOV>"); tokenizer.fit_on_texts(texts)
X=pad_sequences(tokenizer.texts_to_sequences(texts), maxlen=6, padding='post')
model=Sequential([Embedding(100,16,input_length=6),LSTM(16),Dense(1,activation='sigmoid')])
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
model.fit(X,tf.constant(labels),epochs=15,verbose=0)
print("Training accuracy:", model.evaluate(X,labels,verbose=0)[1])
test=pad_sequences(tokenizer.texts_to_sequences(["wonderful movie"]),maxlen=6,padding='post')
print("Positive probability:", float(model.predict(test,verbose=0)[0][0]))
```
