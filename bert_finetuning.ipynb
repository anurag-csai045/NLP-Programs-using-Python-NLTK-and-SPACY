# Experiment 24: Fine-Tuning BERT for Text Classification

**Aim:** Fine-tune a pretrained BERT sequence classifier using Hugging Face Trainer.

**Libraries:** `transformers, datasets, evaluate, torch`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
# Run in Colab or a machine with internet access and sufficient RAM/GPU.
# pip install transformers datasets evaluate accelerate torch
from datasets import Dataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification, TrainingArguments, Trainer
import numpy as np
from sklearn.metrics import accuracy_score, precision_recall_fscore_support
model_name="distilbert-base-uncased"
data=Dataset.from_dict({"text":["I love this product","Amazing experience","Very helpful","I hate this","Awful quality","Very disappointing"],"label":[1,1,1,0,0,0]})
tokenizer=AutoTokenizer.from_pretrained(model_name)
def tokenize(batch): return tokenizer(batch["text"],truncation=True,padding="max_length",max_length=64)
data=data.map(tokenize,batched=True).train_test_split(test_size=.33,seed=42)
model=AutoModelForSequenceClassification.from_pretrained(model_name,num_labels=2)
def metrics(pred):
    y=np.array(pred.label_ids); p=np.argmax(pred.predictions,axis=1)
    precision,recall,f1,_=precision_recall_fscore_support(y,p,average='weighted',zero_division=0)
    return {"accuracy":accuracy_score(y,p),"precision":precision,"recall":recall,"f1":f1}
args=TrainingArguments(output_dir="bert-results",num_train_epochs=2,per_device_train_batch_size=4,per_device_eval_batch_size=4,logging_steps=1,report_to="none")
trainer=Trainer(model=model,args=args,train_dataset=data["train"],eval_dataset=data["test"],compute_metrics=metrics)
trainer.train(); print(trainer.evaluate())
```
