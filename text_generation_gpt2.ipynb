# Experiment 26: Text Generation using GPT-2 on a Custom Dataset

**Aim:** Fine-tune GPT-2 on a tiny text corpus and generate a continuation.

**Libraries:** `transformers, torch`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
# Educational demonstration. Fine-tuning requires internet and compute; use a tiny dataset only for a smoke test.
# pip install transformers torch
from transformers import AutoTokenizer, AutoModelForCausalLM, Trainer, TrainingArguments, DataCollatorForLanguageModeling
from datasets import Dataset
model_name="distilgpt2"
texts=["Natural language processing helps computers work with text.","Machine learning finds patterns in data.","Transformers are useful neural network architectures."]
tokenizer=AutoTokenizer.from_pretrained(model_name); tokenizer.pad_token=tokenizer.eos_token
model=AutoModelForCausalLM.from_pretrained(model_name)
ds=Dataset.from_dict({"text":texts})
def tokenize(batch): return tokenizer(batch["text"],truncation=True,max_length=64)
ds=ds.map(tokenize,batched=True,remove_columns=["text"])
collator=DataCollatorForLanguageModeling(tokenizer=tokenizer,mlm=False)
args=TrainingArguments(output_dir="gpt2-custom",per_device_train_batch_size=2,num_train_epochs=1,report_to="none",save_strategy="no")
trainer=Trainer(model=model,args=args,train_dataset=ds,data_collator=collator)
trainer.train()
inputs=tokenizer("Natural language",return_tensors="pt")
output=model.generate(**inputs,max_new_tokens=30,do_sample=True,temperature=.8,pad_token_id=tokenizer.eos_token_id)
print(tokenizer.decode(output[0],skip_special_tokens=True))
```
