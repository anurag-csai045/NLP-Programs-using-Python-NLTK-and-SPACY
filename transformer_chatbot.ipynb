# Experiment 30: Chatbot using a Pre-trained Transformer

**Aim:** Create a simple conversational loop using a pretrained DialoGPT model.

**Libraries:** `transformers, torch`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
# pip install transformers torch
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
name="microsoft/DialoGPT-small"
tokenizer=AutoTokenizer.from_pretrained(name)
model=AutoModelForCausalLM.from_pretrained(name)
chat_history_ids=None
print("Type 'quit' to stop.")
while True:
    user_text=input("You: ").strip()
    if user_text.lower() in {"quit","exit"}: break
    new_input=tokenizer.encode(user_text+tokenizer.eos_token,return_tensors="pt")
    bot_input=new_input if chat_history_ids is None else torch.cat([chat_history_ids,new_input],dim=-1)
    chat_history_ids=model.generate(bot_input,max_length=min(bot_input.shape[-1]+60,900),pad_token_id=tokenizer.eos_token_id)
    reply=tokenizer.decode(chat_history_ids[:,bot_input.shape[-1]:][0],skip_special_tokens=True)
    print("Bot:",reply)
```
