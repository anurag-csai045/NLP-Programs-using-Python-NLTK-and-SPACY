# Experiment 29: Text Summarization using BART

**Aim:** Summarize a passage with a pretrained BART summarization model.

**Libraries:** `transformers, torch`

**Notes:** Run the cells from top to bottom. Some transformer experiments download pretrained models and require an internet connection; large pretrained embedding files are intentionally not bundled.

```python
# pip install transformers torch
from transformers import pipeline
summarizer=pipeline("summarization",model="facebook/bart-large-cnn")
article=("Natural language processing is a field of artificial intelligence that focuses on interactions "
         "between computers and human language. It includes tasks such as translation, summarization, "
         "sentiment analysis and question answering. Modern transformer models learn contextual patterns "
         "from large text collections and have improved the quality of many language applications. "*3)
result=summarizer(article,max_length=70,min_length=25,do_sample=False)
print("Summary:",result[0]["summary_text"])
```
